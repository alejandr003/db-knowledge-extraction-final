````python
from pathlib import Path
import pandas as pd
from IPython.display import Image, HTML, display, Markdown

PROJECT_ROOT = Path.cwd()
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "clean_dataset.csv"
PLOTS_DIR = PROJECT_ROOT / "assets" / "plots"
DASH_DIR = PROJECT_ROOT / "dashboard" / "screenshots"

df = pd.read_csv(DATA_PATH, parse_dates=["Order Date", "Ship Date"])
print(f"Project root: {PROJECT_ROOT}")
print(f"Rows: {len(df):,} | Columns: {df.shape[1]}")
print(f"Date range: {df['Order Date'].min().date()} -> {df['Order Date'].max().date()}")
````

````python
# Executive KPIs (from processed CSV)
kpis = {
  "Total Sales": df["Sales"].sum(),
  "Total Profit": df["Profit"].sum(),
  "Profit Margin %": df["Profit"].sum() / df["Sales"].sum() * 100,
  "Total Orders": df["Order ID"].nunique(),
  "Total Quantity": int(df["Quantity"].sum()),
  "Average Discount %": df["Discount"].mean() * 100,
  "Loss-Making Rows": int((df["Profit"] < 0).sum()),
  "Loss Rate %": (df["Profit"] < 0).mean() * 100,
  "Loss Amount": df.loc[df["Profit"] < 0, "Profit"].sum(),
  "Avg Shipping Days": df["Shipping Days"].mean(),
}

kpi_df = pd.DataFrame([
  ["Total Sales", f"${kpis['Total Sales']:,.2f}"],
  ["Total Profit", f"${kpis['Total Profit']:,.2f}"],
  ["Profit Margin %", f"{kpis['Profit Margin %']:.2f}%"],
  ["Total Orders", f"{kpis['Total Orders']:,}"],
  ["Total Quantity", f"{kpis['Total Quantity']:,}"],
  ["Average Discount %", f"{kpis['Average Discount %']:.2f}%"],
  ["Loss-Making Rows", f"{kpis['Loss-Making Rows']:,}"],
  ["Loss Rate %", f"{kpis['Loss Rate %']:.2f}%"],
  ["Loss Amount", f"${kpis['Loss Amount']:,.2f}"],
  ["Avg Shipping Days", f"{kpis['Avg Shipping Days']:.2f}"],
], columns=["Metric", "Value"])

kpi_df
````

## 5) Key Findings (Executive Story)

1. **Revenue is concentrated and skewed.** Sales are heavily right-skewed (median much lower than average), so a small share of high-value transactions drives revenue.
2. **Profitability is uneven.** Technology can generate strong profits but also large losses; Furniture carries major margin risk; loss-making sub-categories (e.g., Tables/Bookcases/Supplies) need pricing/cost review.
3. **Discounts are the main profit leak.** Higher discounts are strongly associated with lower profit/margin; discounts above 20% show severe loss rates in the analysis.
4. **Opportunity exists in profitable segments/regions.** Home Office / East-type combinations show stronger margin potential and should be used as a playbook.
5. **Recommendation:** enforce a profitability-based discount policy, review loss-making sub-categories, and focus growth on high-margin segments/regions.
